```python
# Plot a specific trial (you can change the trial number)
if behavioral_data is not None and 'trial' in behavioral_data.columns:
    
    # Choose a specific trial to analyze
    TRIAL_TO_ANALYZE = 5  # Change this to analyze a different trial
    
    available_trials = sorted(behavioral_data['trial'].unique())
    
    if TRIAL_TO_ANALYZE in available_trials:
        print(f"Analyzing trial {TRIAL_TO_ANALYZE}...")
        fig = behavioral_viz.plot_trial_behavioral_data(trial_num=TRIAL_TO_ANALYZE)
        plt.show()
        
        # Get trial statistics
        trial_data = behavioral_data[behavioral_data['trial'] == TRIAL_TO_ANALYZE]
        print(f"\nTrial {TRIAL_TO_ANALYZE} Statistics:")
        print(f"- Duration: {len(trial_data)} samples")
        
        if 'target' in trial_data.columns:
            target_idx = trial_data['target'].iloc[0]
            target_angle = int(target_idx) * 45 if str(target_idx).isdigit() else 'Unknown'
            print(f"- Target: {target_idx} (Direction: {target_angle}°)")
        
        if 'cursor_x' in trial_data.columns and 'cursor_y' in trial_data.columns:
            # Calculate movement metrics
            start_pos = (trial_data['cursor_x'].iloc[0], trial_data['cursor_y'].iloc[0])
            end_pos = (trial_data['cursor_x'].iloc[-1], trial_data['cursor_y'].iloc[-1])
            distance = np.sqrt((end_pos[0] - start_pos[0])**2 + (end_pos[1] - start_pos[1])**2)
            
            print(f"- Start position: ({start_pos[0]:.3f}, {start_pos[1]:.3f})")
            print(f"- End position: ({end_pos[0]:.3f}, {end_pos[1]:.3f})")
            print(f"- Total distance: {distance:.3f}")
            
            # Calculate velocity statistics
            if len(trial_data) > 1:
                vel_x = np.gradient(trial_data['cursor_x'])
                vel_y = np.gradient(trial_data['cursor_y'])
                vel_magnitude = np.sqrt(vel_x**2 + vel_y**2)
                
                print(f"- Peak velocity: {np.max(vel_magnitude):.3f} units/sample")
                print(f"- Mean velocity: {np.mean(vel_magnitude):.3f} units/sample")
                
                # Find movement onset (when velocity exceeds threshold)
                velocity_threshold = 0.1 * np.max(vel_magnitude)  # 10% of peak velocity
                movement_onset = np.where(vel_magnitude > velocity_threshold)[0]
                if len(movement_onset) > 0:
                    print(f"- Movement onset: sample {movement_onset[0]} ({movement_onset[0]/len(trial_data)*100:.1f}% into trial)")
    else:
        print(f"Trial {TRIAL_TO_ANALYZE} not found in data.")
        print(f"Available trials: {available_trials[:20]}...")  # Show first 20 trials
```

```python
# Helper functions for trial analysis
def find_trials_by_target(behavioral_data, target_index):
    """Find all trials for a specific target."""
    if 'target_index' not in behavioral_data.columns:
        return []
    trials = behavioral_data[behavioral_data['target_index'] == target_index]['trial'].unique()
    return sorted(trials)

def get_trial_info(behavioral_data, trial_num):
    """Get detailed information about a specific trial."""
    if 'trial' not in behavioral_data.columns:
        return None
    
    trial_data = behavioral_data[behavioral_data['trial'] == trial_num]
    if len(trial_data) == 0:
        return None
    
    info = {
        'trial_num': trial_num,
        'duration': len(trial_data),
        'target_index': trial_data['target_index'].iloc[0] if 'target_index' in trial_data.columns else 'Unknown',
        'trial_outcome': trial_data['trial_outcome'].iloc[0] if 'trial_outcome' in trial_data.columns else 'Unknown',
        'peak_speed': None,
        'mean_speed': None,
        'movement_onset': None
    }
    
    if 'velocity_x' in trial_data.columns and 'velocity_y' in trial_data.columns:
        # Calculate speed statistics
        speed = np.sqrt(trial_data['velocity_x']**2 + trial_data['velocity_y']**2)
        info['peak_speed'] = np.max(speed)
        info['mean_speed'] = np.mean(speed)
        
        # Find movement onset (10% of peak speed)
        if info['peak_speed'] > 0:
            threshold = 0.1 * info['peak_speed']
            onset_indices = np.where(speed > threshold)[0]
            if len(onset_indices) > 0:
                info['movement_onset'] = onset_indices[0]
    
    return info

def analyze_target_direction(behavioral_data, target_index):
    """Analyze all trials for a specific target direction."""
    trials = find_trials_by_target(behavioral_data, target_index)
    
    if len(trials) == 0:
        print(f"No trials found for target {target_index}")
        return []
    
    # Get number of targets to calculate direction
    n_targets = 8  # Default
    if 'num_targets' in behavioral_data.columns:
        n_targets = behavioral_data['num_targets'].iloc[0]
    
    direction = target_index * (360 / n_targets)
    
    print(f"Target {target_index} Analysis:")
    print(f"- Direction: {direction}°")
    print(f"- Number of trials: {len(trials)}")
    
    # Analyze trial characteristics
    durations = []
    peak_speeds = []
    outcomes = []
    
    for trial in trials:
        info = get_trial_info(behavioral_data, trial)
        if info:
            durations.append(info['duration'])
            if info['peak_speed'] is not None:
                peak_speeds.append(info['peak_speed'])
            outcomes.append(info['trial_outcome'])
    
    if durations:
        print(f"- Mean duration: {np.mean(durations):.1f} ± {np.std(durations):.1f} samples")
    if peak_speeds:
        print(f"- Mean peak speed: {np.mean(peak_speeds):.3f} ± {np.std(peak_speeds):.3f} units/s")
    
    # Show outcome distribution
    from collections import Counter
    outcome_counts = Counter(outcomes)
    print(f"- Trial outcomes: {dict(outcome_counts)}")
    
    return trials

# Example usage
if behavioral_data is not None and 'trial' in behavioral_data.columns:
    print("Available helper functions (updated for velocity data):")
    print("1. find_trials_by_target(behavioral_data, target_index)")
    print("2. get_trial_info(behavioral_data, trial_num)")
    print("3. analyze_target_direction(behavioral_data, target_index)")
    
    # Get available targets
    if 'target_index' in behavioral_data.columns:
        available_targets = sorted(behavioral_data['target_index'].unique())
        print(f"\nAvailable targets: {available_targets}")
        
        # Example: Analyze first available target
        if len(available_targets) > 0:
            first_target = available_targets[0]
            print(f"\nExample: Analyzing target {first_target}...")
            target_trials = analyze_target_direction(behavioral_data, first_target)
            
            if len(target_trials) > 0:
                print(f"\nFirst few trials for target {first_target}: {target_trials[:5]}")
                
                # Plot the first trial for this target
                print(f"\nPlotting first trial for target {first_target}: Trial {target_trials[0]}")
                fig = behavioral_viz.plot_trial_behavioral_data(trial_num=target_trials[0])
                plt.show()
    
    print("\n" + "="*50)
    print("Updated functions work with velocity data and trial segmentation!")
    print(f"Example: analyze_target_direction(behavioral_data, {available_targets[1] if len(available_targets) > 1 else 0})")
    print("Example: get_trial_info(behavioral_data, 5)  # for trial 5")
```
